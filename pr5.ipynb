```python
f = open("p1.txt", "w")
f.write("Hello Python")
f.close()

f = open("p1.txt", "a")
f.write("\nthis is a text file")
f.close()

f = open("p1.txt", "r")
print(f.read())
f.close()
```

```python
import csv
f = open("p2.csv", "w", newline="")
w = csv.writer(f)
w.writerow(["Name", "Marks"])
w.writerow(["Reema", 95])
f.close()

f = open("p2.csv", "a", newline="")
w = csv.writer(f)
w.writerow(["dia", 80])
f.close()

f = open("p2.csv", "r")
for row in csv.reader(f):
    print(row)
f.close()
```
