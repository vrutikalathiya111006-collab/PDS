```python
import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv("employee.csv")

# Use first five columns only
data = df.iloc[:, :5]

print("First Five Columns:")
print(data.head())

# Statistical measures for Age
print("\n--- Statistical Measures for Age ---")
print("Mean:", data["Age"].mean())
print("Median:", data["Age"].median())
print("Mode:", data["Age"].mode()[0])
print("Variance:", data["Age"].var())
print("Standard Deviation:", data["Age"].std())
print("Minimum:", data["Age"].min())
print("Maximum:", data["Age"].max())

# One sample t-test
print("\n--- One Sample T-Test ---")
t_age = stats.ttest_1samp(data["Age"], 35)
print("Age T-test:", t_age)

```
