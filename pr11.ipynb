```python
import pandas as pd

# Read CSV file
df = pd.read_csv("employee.csv")

# Use only first five columns
df = df.iloc[:, :5]

print("First 5 Columns:")
print(df.head())

# Create a Series
age = df["Age"]

print("\nSeries - Age:")
print(age.head())

# Data Selection
print("\nSelected Columns:")
print(df[["Age", "Department"]].head())

# Filtering
print("\nEmployees with Age greater than 40:")
print(df[df["Age"] > 40].head())

# Grouping
print("\nEmployees grouped by Department:")
print(df.groupby("Department")["DailyRate"].mean())

# Aggregation
print("\nAggregation by Department:")
print(df.groupby("Department")["DailyRate"].agg(["count", "mean", "max", "min"]))
```
